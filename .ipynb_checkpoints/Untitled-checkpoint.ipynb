```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram,linkage
```

```python
countrys=pd.read_csv('Country-data.csv')
```

```python
countrys.head()
```

```python
countrys.groupby(by=['country'])['income'].sum()
```

```python
countrys.shape
```

```python
countrys.isna().sum()
```

```python
x=countrys.drop(columns=['country'])
```

```python
x.columns
```

```python
scalar=StandardScaler()
x_scaled=scalar.fit_transform(x)
```

## dendogram

```python
linked=linkage(x_scaled,method='ward')
```

```python

dendrogram(linked, leaf_rotation=90, leaf_font_size=8)
plt.title('Dendrogram for Country Data (Ward Linkage)')
plt.xlabel('Countries')
plt.ylabel('Euclidean Distance')
plt.axhline(y=12, color='r', linestyle='--', label='Threshold Line (3 Clusters)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
model=AgglomerativeClustering(n_clusters=3,metric='euclidean',linkage='ward')
```

```python
countrys['cluster']=model.fit_predict(x_scaled)
```

```python
print(countrys.groupby('cluster').mean(numeric_only=True))
```

```python
developed=countrys.loc[countrys['cluster']==0,['country']]
developed
```

```python
developing_country=countrys.loc[countrys['cluster']==1,['country']]
developing_country.values
```

```python
undeveloped_country=countrys.loc[countrys['cluster']==2,['country']]
undeveloped_country
```

## visualize the cluster

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='gdpp', 
    y='child_mort', 
    hue='cluster', 
    data=countrys, 
    palette='viridis', 
    s=100
)
plt.title('Country Clusters Visualisation (GDP vs Child Mortality)')
plt.xlabel('GDP per capita')
plt.ylabel('Child Mortality')
plt.show()
```
